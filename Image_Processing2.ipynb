```python
# cv2.flip(görüntünüz,döndürme_parametresi)
# cv2'nin flip fonksiyonu ile görüntüyü yata, dikey ve hem yatay hem de dikey
# eksenlerde çevirebiliyoruz
# iki parametre alır, ilk olarak görüntünün kendisi, ikinci olarak da 
# döndüreceğiniz eksen
# döndürme ekseni:0 = dikey eksen
# döndürme ekseni:1 = yatay eksen
# döndürme ekseni:-1 = hem dikey hem de yatay eksen
```

```python
my_rgb_vertical_img = cv2.flip(my_rgb_img,0)
my_rgb_horizontal_img = cv2.flip(my_rgb_img,1)
my_rgb_both_img = cv2.flip(my_rgb_img,-1)
```

```python
# 10x10'luk boş bir figür oluşturuyoruz
fig = plt.figure(figsize=(10,10))

# 2 satır ve 2 sütundan oluşan 4 görüntünün eşit şekilde sığacağı subplot'ları
# yani alt figürleri oluşturuyoruz

plt.subplot(2,2,1)
plt.imshow(my_rgb_img)
plt.axis("off")
plt.title("Original RGB Image")

plt.subplot(2,2,2)
plt.imshow(my_rgb_vertical_img)
plt.axis("off")
plt.title("Vertical Image")

plt.subplot(2,2,3)
plt.imshow(my_rgb_horizontal_img)
plt.axis("off")
plt.title("Horizontal Image")

plt.subplot(2,2,4)
plt.imshow(my_rgb_both_img)
plt.axis("off")
plt.title("Vertical and Horizontal Image")
```

```python
# Sıfırdan, kendi flip fonksiyonumuzu yazıyoruz.

# Aşama-1: cv2.flip() fonksiyonu 2 parametreli olduğu için biz de kendi
# fonksiyonumuzu tanımlarken 2 parametreli bir fonksiyon tanımlıyoruz

# Aşama-2: ilk parametre, dönüştürülmek istenen orijinal görüntü (numpy.ndarray)
# ikinci parametrede 0,1 ya da -1 parametresi alan bir parametre olacak
# ilk parametrenin tipi numpy.ndarray, 2. parametrenin tipi int

def my_flip(org_img,flip_number):
    if flip_number==0:
        org_img = org_img[::-1,:,:]
    elif flip_number==1:
        org_img = org_img[:,::-1,:]
    elif flip_number==-1:
        org_img = org_img[::-1,::-1,:]
    else:
        print("Lütfen doğru bir parametre giriniz")
    return org_img
```

```python
def my_flip(org_img,flip_number):
    if flip_number==0:
        org_img=org_img[::-1,:,:]
    elif flip_number==1:
        org_img=org_img[:,::-1,:]
    elif flip_number==-1:
        org_img=org_img[::-1,::-1,:]
    else: 
        print("lütfen dogru parametre giriniz")
    return org_img

my_flip_0=my_flip(my_rgb_img,0)
my_flip_1=my_flip(my_rgb_img,1)
my_flip_2=my_flip(my_rgb_img,-1)
```
