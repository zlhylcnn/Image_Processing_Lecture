```python
# Binary Thresholding: belirlenen eşik değerine göre (thresh value),
# orijinal görüntünün pikseli, bu eşik değerinden küçükse siyah rengine
# dönüştürülür yani 0 değerini alır, ilgili piksel eşik değerinden büyükse
# beyaz rengine dönüştürülür yani 255 değerini alır.

# Binary Inverse (Binary_inv) Thresholding: Binary thresholding'in tam tersidir.
# ilgili piksel, eşik değerinden küçükse 255 değerini alır (beyaz rengine
# dönüştürülür), eşik değerinden büyükse 0 değerini alır (siyah rengine
# dönüştürülür.

# Trunc Thresholding: Görüntünün parlak renklerinin kısılması için kullanılır.
# Şöyle ki, ilgili piksel değeri eşik değerinden büyükse eşik değerine eşitlenir
# eşik değerinden küçükse aynı bırakılır.

# TOZERO Thresholding: eğer ilgili piksel değeri, eşik değerinden büyükse, olduğu
# gibi korunur yani değişmez aynı kalır. Yok eğer, piksel değerinden küçükse,
# 0 yapılır yani siyah rengine dönüşür.

# TOZERO Thresholding: TOZERO'nun tam tersidir.
# eğer ilgili piksel değeri, eşik değerinden büyükse, 
# 0 yapılır yani siyah rengine dönüşür.
# Eğer küçükse olduğu gibi korunur yani değişmez aynı kalır.
```

```python
# Python'da 2 adet cv2.resize() fonksiyonu yazınız.
# Bu fonksiyonlar 1 adet parametre alacak. O da görüntünün kendisi olacak 
# Görüntüyü büyütürken de küçültürken de genişlik ve yüksekliğinin
# %25'i kadar olacaktır. Her iki fonksiyon da geriye, işlenmiş görüntünün
# kendisini döndürecektir.

# cv2.resize() kullanımı
# cv2.resize(original_img,(genişlik,yükseklik),cv2.INTER_LANCZOS4)
```

```python
def img_zoom_in(img_org):
    # Görüntünün satırının (yüksekliğinin) %25'ini hesapladık
    new_height = img_org.shape[0] + int(img_org.shape[0]*0.25)
    new_weight = img_org.shape[1] + int(img_org.shape[1]*0.25)

    new_img = cv2.resize(img_org,(new_weight, new_height),cv2.INTER_LANCZOS4)
    return new_img

def img_zoom_out(img_org):
    # Görüntünün satırının (yüksekliğinin) %25'ini hesapladık
    new_height = img_org.shape[0] - int(img_org.shape[0]*0.25)
    new_weight = img_org.shape[1] - int(img_org.shape[1]*0.25)

    new_img = cv2.resize(img_org,(new_weight, new_height),cv2.INTER_LANCZOS4)
    return new_img
```

```python
my_rgb_img = cv2.cvtColor(my_img, cv2.COLOR_BGR2RGB)
```

```python
new_big_img = img_zoom_in(my_rgb_img)
new_small_img = img_zoom_out(my_rgb_img)

print(f"Orijinal Görüntü = {my_rgb_img.shape[:2]}")
print(f"Büyütülmüş Görüntü = {new_big_img.shape[:2]}")
print(f"Küçültülmüş Görüntü = {new_small_img.shape[:2]}")
```

```python
fig = plt.figure(figsize=(10,7))

plt.subplot(1,3,1)
plt.imshow(my_rgb_img)
plt.axis("off")
plt.title("Original Gray Image")

plt.subplot(1,3,2)
plt.imshow(new_big_img)
plt.axis("off")
plt.title("BIGGER Image")

plt.subplot(1,3,3)
plt.imshow(new_small_img)
plt.axis("off")
plt.title("SMALLER Image")

plt.tight_layout()
plt.savefig("./Image_Resizing_with_automatic_calculation.png",dpi=300)
```
