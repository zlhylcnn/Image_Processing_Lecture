```python
import cv2
import numpy as np
```

```python
model_pathh = 'haarcascade_frontalface_default.xml'
#model_pathh: Kullanılacak Haar Cascade modelinin dosya adını tutar. haarcascade_frontalface_default.xml modeli, tipik olarak önden çekilmiş insan yüzlerini tanımak için eğitilmiş ağırlıkları içerir.
#cv2.CascadeClassifier(...): Bu sınıf, yüklenen XML dosyasındaki yüz tanıma algoritmasının kurallarını/ağırlıklarını yükler ve bu ağırlıklar artık face_cascades değişkeni aracılığıyla kullanılır.

face_cascades = cv2.CascadeClassifier(cv2.data.haarcascades + model_pathh)

my_bgr_img = cv2.imread("./deneme.jpg")
my_gray_img = cv2.imread("./deneme.jpg",0)#Orijinal görüntüyü gri tonlamalı formatta yükler. Haar Cascade tabanlı yüz tanıma algoritmaları, hesaplama verimliliği ve performansı için daima gri tonlamalı görüntü üzerinde çalışır.
```

```python
# Ağırlıklara sahip modelimizin (face_cascades), bir yüzü tanımlarken
# aşağıdaki parametrelere ait değerlere (kurallara) bağlı kalmasını sağladık

faces = face_cascades.detectMultiScale(#faces Çıktısı: Fonksiyon, görüntüde tespit ettiği her bir yüz için bir liste döndürür. Listedeki her eleman, tespit edilen yüzün konumunu ve boyutunu (x, y, w, h) formatında tutan bir NumPy dizisidir.
    my_gray_img,# işlenecek görüntü (daima gri seviye görüntü)
    scaleFactor = 1.1,#görüntünün yakınlaştırma derecesi
    minNeighbors = 5,# yanlış pozitifleri ayrıştırmak için
    minSize = (5,5)#en küçük yüz boyutu#minSize = (5,5): Tespit edilecek en küçük yüz boyutudur. Bu değerden küçük olan yüzler yok sayılır.
)

sayac = 0

for (x,y,w,h) in faces:#x, y: Yüzün bulunduğu dikdörtgenin sol üst köşe koordinatları.#w, h: Yüzün genişliği ve yüksekliği.
    cv2.rectangle(my_bgr_img,(x,y),(x+w,y+h),(0,255,0),2)#x,y): Dikdörtgenin sol üst köşesi.#(x+w, y+h): Dikdörtgenin sağ alt köşesi.
    sayac += 1
    #(0,255,0): Çizgi rengi (Yeşil).#2: Çizgi kalınlığı.#sayac: Kaç adet yüz tespit edildiğini sayar.

print(f"{sayac} adet futbolcu tespit edildi!!!")
cv2.imshow("Futbolcular", my_bgr_img)
cv2.waitKey(0)
cv2.destroyAllWindows()
#Özetle: Kod, bir görüntü yükler, yüz tanıma modelini hazırlar, görüntüde yüzleri arar ve bulduğu her yüzün etrafına yeşil bir kutu çizerek sonucu ekranda gösterir.
```
