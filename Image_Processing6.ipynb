```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
```

```python
my_bgr_img = cv2.imread("./bonus.jpg")
my_gray_img = cv2.cvtColor(my_bgr_img,cv2.COLOR_BGR2GRAY)
```

```python
print(type(my_bgr_img))
print(type(my_gray_img))
```

```python
# ilk iş olarak, gri seviyedeki görüntünün eşiğini hesaplayarak binary
# hale getiriyoruz
# bu sayede, kenar tespitini en başarılı şekilde yapıp, bu kenarların
# birleşmesi sonucu kontoru bulmayı hedefliyoruz

my_bgr_img = cv2.imread("./bonus.jpg")
my_gray_img = cv2.imread("./bonus.jpg",0)

ret, thresh = cv2.threshold(my_gray_img,127,255,cv2.THRESH_BINARY)

# ret olarak thresh değeri dönüyor yani 127
# thresh değişkeni de eşiklenmiş görüntüyü saklıyor

# eşiklenmiş görüntüyü elde ettikten sonra konturları bulmaya başlıyoruz
# kontur bulma işleminde de 2 adet değer dönüyor. Konturların hiyerarşisi
# (anne/çocuk ilişkisi) ve konturların listesi. Bize lazım olan değer,
# konturların listesi, yani dönen ilk değişken

contour, hierarchy = cv2.findContours(thresh, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)

# eşiklenmiş görüntünün (thresh değşkeni) konturlarını bulurken
# RETRive external metodunu kullan, yani sadece dış kenarları al
# bunları zincir haline getirirken de (birleştirme işleminde de)
# CHAIN_APPROX_SIMPLE metodunu kullan diyoruz.

cv2.drawContours(my_bgr_img, contour,-1,(0,255,0),2)

# drawContours fonksiyonu, bir önceki aşamada elde edilen konturların listesini
# alır, onu renkli görüntü üzerine uygular (koordinat sistemiyle)
# ve orijinal görüntü artık, üzerindeki konturların (kenarların bir zincirle
# birleştirilmesi sonucunda) elde edilmesiyle bitirilir.

# fonksiyonun parametreleri: renkli_görüntü_değişkeni, kontur_listesi,
# -1: konturların tamamını uygula (kesintisiz-hepsi)
# (0,255,0) konturları gösterirken yeşil rengini seç (istediğiniz rengi verebilirsiniz)
# 2: konturları gösteren rengin kalınlık derecesi

cv2.imshow("Image's Contour",my_bgr_img)
cv2.imshow("Thresh Image",thresh)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
my_bgr_img = cv2.imread("./bonus.jpg")
my_gray_img = cv2.imread("./bonus.jpg",0)#cv2.imread("./bonus.jpg", 0): Bu ikinci yöntem, görüntüyü doğrudan gri tonlamalı olarak yükler (ikinci parametre 0 gri tonlamayı belirtir).

ret, thresh = cv2.threshold(my_gray_img,127,255,cv2.THRESH_BINARY_INV + cv2.THRESH_OTSU)#Çıkış: İki şey döndürür:ret: Fonksiyonun hesapladığı veya kullandığı eşik değeri. thresh: İşlem sonucunda elde edilen yeni ikili (binary) görüntü.
#+ cv2.THRESH_OTSU: Bu bayrak, OpenCV'ye manuel eşik değeri ($127$) yerine Otsu Metodunu kullanmasını söyler.Otsu Metodu: Görüntüdeki siyah ve beyaz piksellerin dağılımını (histogramını) analiz eder ve arka planı ön plandan en iyi şekilde ayıracak eşik değerini (örneğin $105$ veya $180$ gibi) otomatik olarak hesaplar.
#my_gray_img üzerindeki nesne/arka plan ayrımını Otsu metoduyla otomatik olarak bul, ancak ikili görüntüyü ters çevir; yani koyu olanı beyaz, parlak olanı siyah yap.


thresh = cv2.morphologyEx(thresh, cv2.MORPH_CLOSE, np.ones((3,3),dtype=np.uint8),iterations=2)#cv2.morphologyEx(...): Bu, OpenCV'deki gelişmiş morfolojik operatörleri uygulayan genel bir fonksiyondur.
#thresh: Üzerinde işlem yapılacak giriş görüntüsüdür. Bu, bir önceki adımda Otsu ve Ters İkili Eşikleme ile elde edilmiş olan siyah-beyaz (binary) görüntüdür.
#cv2.MORPH_CLOSE: Uygulanacak morfolojik işlemin tipidir: Kapatma (Closing).Kapatma Nedir? Kapatma işlemi, önce genişletme (Dilation), ardından aşındırma (Erosion) işlemlerini sırayla uygular.
#Amacı: Nesnelerin içindeki küçük delikleri doldurmak, nesne sınırları üzerindeki küçük kesintileri birleştirmek ve konturları yumuşatmaktır. Kontur tespiti öncesinde görüntüyü temizlemek için idealdir.

#np.ones((3,3), dtype=np.uint8): Morfolojik işlemin nasıl uygulanacağını belirleyen çekirdek (kernel) matrisidir.np.ones((3,3)): $3 \times 3$ boyutlarında, içi tamamen $1$'lerden oluşan bir matris oluşturur.Anlamı: İşlem, görüntünün her pikseline $3 \times 3$'lük komşuluk alanında uygulanacaktır. Bu, küçük ve lokal iyileştirmeler için standart bir boyuttur.
#iterations=2: Morfolojik Kapatma işleminin iki kez tekrarlanacağını belirtir.
#Amacı: Tek bir uygulamanın yeterli olmadığı durumlarda, temizleme ve iyileştirme etkisini artırmak için işlem yinelenir.
#Satırın Nihai Etkisi
#Bu satır, Ters Otsu eşiklemesiyle elde edilen ikili görüntüdeki nesne hatlarını:
#Daha düzgün hale getirir.
#Küçük gürültüleri veya boşlukları doldurur.
#Böylece, bu işlemden sonra çalıştırılan cv2.findContours() fonksiyonu, daha temiz ve birleşik kenar zincirleri (konturlar) bulabilir. Bu, görüntüdeki nesnelerin sınırlarını doğru bir şekilde temsil etmek için kritik bir ön işlemdir.


contour, hierarchy = cv2.findContours(thresh, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)#Bu satır, nesne sınırlarını (konturlarını) tespit eder ve konturları bir liste olarak döndürür.
#thresh: Morfolojik işlemden geçmiş (yani iyileştirilmiş) ikili (siyah-beyaz) görüntü.
#contour: Bulunan tüm konturların bir listesi (bize lazım olan temel çıktı budur). Her kontur, bir NumPy dizisidir ve kontur üzerindeki tüm koordinat noktalarını içerir.
#hierarchy: Konturların birbirleriyle olan hiyerarşik ilişkisini (hangi konturun dış kontur, hangisinin iç kontur olduğu gibi) gösteren bir matris.
#cv2.RETR_EXTERNAL: Kontur Bulma Modu. Yalnızca en dıştaki (yani nesnelerin ana sınırlarını) konturları alır. Nesnenin içindeki deliklerin konturları yoksayılır.
#cv2.CHAIN_APPROX_SIMPLE: Yaklaşım Metodu. Kontur üzerindeki gereksiz tüm noktaları kaldırır ve konturu tanımlayan sadece başlangıç/bitiş noktalarını veya köşe noktalarını saklar. Bu, bellekten tasarruf sağlar.
#contour: Bir önceki adımda bulunan konturların listesi.

#-1: Kontur İndeksi. Listenin tamamındaki tüm konturların çizileceğini belirtir. (Örneğin sadece ilk konturu çizmek için 0 kullanılırdı).
#(0,255,0): Renk. Konturların çizileceği BGR renk kodu. Bu kod Yeşil rengi temsil eder.
#2: Kalınlık. Kontur çizgisinin piksel cinsinden kalınlığıdır.
cv2.drawContours(my_bgr_img, contour,-1,(0,255,0),2)

cv2.imshow("Image's Contour",my_bgr_img)
cv2.imshow("Thresh Image",thresh)
cv2.waitKey(0)
cv2.destroyAllWindows()

#Bu son kısım, OpenCV'nin görüntüleri bir pencerede göstermek için kullandığı standart kapanış kodudur:

#cv2.imshow("Image's Contour", my_bgr_img): Konturları çizilmiş olan renkli görüntüyü "Image's Contour" başlıklı bir pencerede gösterir.

#cv2.imshow("Thresh Image", thresh): Siyah-beyaz ikili görüntüyü "Thresh Image" başlıklı bir pencerede gösterir.

#cv2.waitKey(0): Bir tuşa basılmasını bekler. 0 sonsuza kadar bekleneceği anlamına gelir. Bu, kullanıcı kapatana kadar pencerelerin açık kalmasını sağlar.

#cv2.destroyAllWindows(): Herhangi bir tuşa basıldıktan sonra açılan tüm OpenCV pencerelerini kapatır.

#Özetle: Kod, önce nesne sınırlarını matematiksel olarak tespit eder, ardından bu sınırları orijinal görüntünün üzerine yeşil çizgilerle çizer ve son olarak hem ikili görüntüyü hem de sonuç görüntüyü ekranda gösterir.
```
